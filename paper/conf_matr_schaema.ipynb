```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

rng = np.random.default_rng(42)
n_classes = 5
labels = ['A', 'B', 'C', 'D', 'E']

# Schematic matrix — strong diagonal with realistic off-diagonal noise
conf = rng.integers(0, 4, size=(n_classes, n_classes))
np.fill_diagonal(conf, rng.integers(18, 25, size=n_classes))
conf = conf.astype(float)
# Normalise rows to [0,1]
conf_norm = conf / conf.sum(axis=1, keepdims=True)

fig, ax = plt.subplots(figsize=(2.5, 2.5))

im = ax.imshow(conf_norm, cmap='viridis', vmin=0, vmax=1, aspect='equal')

# Minimal tick labels — small font, rotated
ax.set_xticks(range(n_classes))
ax.set_yticks(range(n_classes))
ax.set_xticklabels(labels, fontsize=20, ha='center')
ax.set_yticklabels(labels, fontsize=20)

ax.set_xlabel('PRED', fontsize=20, labelpad=3)
ax.set_ylabel('TRUE', fontsize=20, labelpad=3)
ax.tick_params(length=0)  # hide tick marks, keep labels

# Thin border
for spine in ax.spines.values():
    spine.set_linewidth(0.5)

plt.tight_layout(pad=0.3)
plt.savefig('conf_matrix_schematic.png', dpi=300, bbox_inches='tight')
plt.show()
```
